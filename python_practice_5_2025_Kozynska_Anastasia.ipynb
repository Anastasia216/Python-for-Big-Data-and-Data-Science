Визначте метод get_volume в кожному з них так, щоби вони розраховували об'єм у залежності від формули для кожного типу фігури. Створіть обʼєкт класу Циліндр та обʼєкт класу Куб. Виконайте операцію суми за допомогою оператора + між ними, аби знайти сумарний об'єм.

```python
from math import pi
class Cylinder(Figure):
  def __init__(self, radius, height):
    self.radius = radius
    self.height = height
  def get_volume(self):
    volume = pi * self.radius**2 * self.height
    return volume
class Cube(Figure):
  def __init__(self, side):
    self.side = side
  def get_volume(self):
    volume = self.side**3
    return volume
cylinder_obj = Cylinder(3,5)
cube_obj = Cube(4)
print(cube_obj + cylinder_obj)
```

7. (3б). Продемонструйте різницю між isinstance та issubclass.

```python
class Vehicle:
    pass

class Car(Vehicle):
    pass

class Bike(Vehicle):
    pass

my_car = Car()
my_bike = Bike()


# isinstance — перевіряє об'єкт
print(isinstance(my_car, Car)) # True
print(isinstance(my_car, Vehicle)) # True
print(isinstance(my_car, Bike)) # False
print(isinstance(my_bike, Vehicle)) # True

# issubclass — перевіряє класи
print(issubclass(Car, Vehicle)) # True
print(issubclass(Bike, Vehicle)) # True
print(issubclass(Vehicle, Car)) # False
print(issubclass(Car, Car)) # True
```

8. (13б). Створіть клас EWallet з приватними атрибутами balance та id. Реалізуйте методи поповнення та зняття коштів, забезпечивши належну інкапсуляцію. Підказка: використовуйте декоратори property та setter.

```python
class EWallet:
  def __init__ (self, balance, id):
    self._balance = balance
    self._id = id
  @property
  def balance (self):
    return self._balance
  @balance.setter
  def balance(self, value):
     if value < 0:
      raise ValueError("balance cannot be negative")
     self._balance = value
wallet = EWallet(15,1)
wallet._id
wallet.balance = -1
```

9. (11б). Створіть клас ShoppingCart, який містить список об'єктів типу Item. Реалізуйте методи для додавання айтемів, їх видалення та відображення списку айтемів.

```python
class Item:
  def __init__ (self,name):
    self.name = name

class ShoppingCart:
  def __init__ (self, items):
    self.items = items
  def print (self):
    res = ""
    for item in self.items:
      res += f"{item.name},"
    print(res)

item_list = [Item("flour"), Item("milk"), Item("sugar")]
cart = ShoppingCart(item_list)
print(cart)
cart.print()
```

Ітерабельний об’єкт — це «колекція», з якої можна отримати ітератор (наприклад, рядок "abcd"), а ітератор — це об’єкт, який пам’ятає поточну позицію і дозволяє по черзі отримувати елементи за допомогою next().

13. (11б). Створіть клас, що має визначений Iterator Protocol та при кожному виклику next повертає назву дня тижня, поки вони не вичерпаються. Зверніть увагу, що порядок має значення.

```python
class WeekDaysIterator:
    def __init__(self):
        self.days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
        self.index = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.index < len(self.days):
            day = self.days[self.index]
            self.index += 1
            return day
        else:
            raise StopIteration

week = WeekDaysIterator()

print(next(week))
print(next(week))
```
